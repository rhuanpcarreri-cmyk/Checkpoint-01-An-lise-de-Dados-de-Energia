```python
#Determine o consumo máximo registrado em cada uma das três zonas.
CMAX1 = dados['Consumo_Zona_1'].max()
print(f"O maior consumo registrado na zona 1 é {CMAX1:.2F} kwh")
```

```python
#Determine o consumo máximo registrado em cada uma das três zonas.
CMAX2 = dados['Consumo_Zona_2'].max()
print(f"O maior consumo registrado na zona 2 é {CMAX2:.2F} kwh")
```

```python
#Determine o consumo máximo registrado em cada uma das três zonas.
CMAX3 = dados['Consumo_Zona_3'].max()
print(f"O maior consumo registrado na zona 3 é {CMAX3:.2F} kwh")
```

```python
#Identifique qual zona apresenta o maior pico de consumo na amostra.
#Identifique qual zona apresenta o maior pico de consumo na amostra.
consumos_max = {
    'Zona 1': CMAX1,
    'Zona 2': CMAX2,
    'Zona 3': CMAX3
}

zona_maior_pico = max(consumos_max, key=consumos_max.get)
valor_maior_pico = consumos_max[zona_maior_pico]

print(f"A {zona_maior_pico} apresenta o maior pico de consumo, com {valor_maior_pico:.2f} kwh")
```

```python
#para a zona identificada, calcule 70% do valor máximo e crie um DataFrame com os registros acima desse limiar.
P70 = 0.70 * dados['Consumo_Zona_1'].max()
print(f"A limiar que corresponde em 70% do valor máximo é de {P70:.2f} kWh")
```

```python
#crie um DataFrame com os registros acima desse limiar.
#Conte os registros e calcule o percentual que eles representam.

df_70 = dados[dados['Consumo_Zona_1'] > P70]

total_registros = dados.shape[0]
percentual = (df_70.shape[0] / total_registros) * 100
print(f"Os registros selecionados representam {percentual:.2f}% da amostra. E eles são um total de {df_70.shape[0]}")
```

```python
#Calcule a temperatura média e crie um segundo DataFrame contendo consumo acima do limiar e temperatura acima da média.
media_temperatura = dados['Temperature'].mean()
print(f"A temperatura média é de {media_temperatura:.2f} °C")

df_70_temp_acima_media = df_70[df_70['Temperature'] > media_temperatura]

print(f"\nQuantidade de registros no df_70 (consumo acima do limiar): {df_70.shape[0]}")
print(f"Quantidade de registros no df_70_temp_acima_media (consumo acima do limiar E temperatura acima da média): {df_70_temp_acima_media.shape[0]}")

print("Após a inclusão da condição ambiental (temperatura acima da média), a quantidade de registros foi reduzida, pois estamos aplicando um filtro adicional, selecionando apenas os casos em que ambas as condições (consumo alto e temperatura alta) são atendidas.")
```

Isso ocorre porque potência ativa e corrente são fortemente correlacionadas (P = V×I), com a tensão relativamente estável — logo, valores altos de potência já implicam valores altos de corrente, tornando a segunda condição redundante nesse conjunto de dados.

```python
#Calcule separadamente 70% do máximo solar e 70% do máximo eólico.

S70 = SMAX * 0.7
print(f"70% do máximo registrado para energia solar é {S70:.2f} MWh")

E70 = EMAX * 0.7
print(f"70% do máximo registrado para energia eólica é {E70:.2f} MWh")
```

```python
#Crie um DataFrame para registros de alta geração solar e outro para registros de alta geração eólica.
#Conte os registros de cada DataFrame e calcule seus respectivos percentuais.
df_70_S = dados[dados['Geracao_Solar'] > S70]

total_registros = dados.shape[0]
percentual1 = (df_70_S.shape[0] / total_registros) * 100
print(f"Os registros selecionados representam {percentual1:.2f}% da amostra e eles são um total de {df_70_S.shape[0]} registros")


df_70_E = dados[dados['Geracao_Eolica'] > E70]

total_registros = dados.shape[0]
percentual2 = (df_70_E.shape[0] / total_registros) * 100
print(f"Os registros selecionados representam {percentual2:.2f}% da amostra e eles são um total de {df_70_E.shape[0]} registros")
```

```python
#Compare qual fonte aparece com maior frequência acima de 70% do seu próprio máximo.
if df_70_S.shape[0] > df_70_E.shape[0]:
    print("A geração solar aparece com maior frequência acima de 70% do seu máximo.")
elif df_70_E.shape[0] > df_70_S.shape[0]:
    print("A geração eólica aparece com maior frequência acima de 70% do seu máximo.")
else:
    print("Ambas as fontes aparecem com a mesma frequência acima de 70% do seu máximo.")
```

Explique por que não seria adequado utilizar diretamente o mesmo valor numérico de potência como limite para as duas fontes sem antes observar a escala de cada variável:

Porque solar e eólica podem ter escalas de produção bem diferentes (uma pode variar entre valores bem menores e outra atingir picos bem maiores). Se você usar o mesmo número fixo como limite para as duas, ele pode ser adequado (ex: representar um pico razoável) para uma fonte, mas ser exagerado ou insignificante para a outra — classificando errado o que é "alto" ou "baixo" consumo/produção em cada caso.

Por isso, antes de definir um limite, é preciso observar a distribuição (mínimo, máximo, média) de cada variável separadamente, para que o limite escolhido faça sentido dentro da escala real de cada fonte.
